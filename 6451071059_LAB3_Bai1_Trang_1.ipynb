```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU B: KHẢO SÁT BỘ DỮ LIỆU IRIS")
print("=" * 60 + "\n")

import pandas as pd
from sklearn.datasets import load_iris

# Load lại dữ liệu để chạy độc lập
iris = load_iris()

# 1. In ra tên các thuộc tính, thuộc tính phân lớp
print("1. TÊN CÁC THUỘC TÍNH VÀ THUỘC TÍNH PHÂN LỚP:")
print("- Tên các thuộc tính (feature_names):")
print(iris.feature_names)
print("- Thuộc tính phân lớp (target_names):")
print(iris.target_names)
print("-" * 50)

# 2. In ra số lượng mẫu, số lượng thuộc tính
print("\n2. SỐ LƯỢNG MẪU VÀ SỐ LƯỢNG THUỘC TÍNH:")
print(f"- data.shape: {iris.data.shape}")
print(f"  + Số lượng mẫu (samples): {iris.data.shape[0]}")
print(f"  + Số lượng thuộc tính (features): {iris.data.shape[1]}")
print(f"- target.shape: {iris.target.shape}")
print("-" * 50)

# 3. In ra 5 mẫu dữ liệu đầu tiên
print("\n3. 5 MẪU DỮ LIỆU ĐẦU TIÊN:")
print("- 5 mẫu dữ liệu đặc trưng (data[:5]):")
print(iris.data[:5])
print("- 5 nhãn tương ứng (target[:5]):")
print(iris.target[:5])
print("-" * 50)

# 4. Thống kê count, mean, std, min, max, tứ vị phân của bộ dữ liệu
print("\n4. BẢNG THỐNG KÊ (DESCRIBE):")
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print(df_iris.describe())
```

```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU C: CHUẨN HÓA DỮ LIỆU VỀ ĐOẠN [0, 1]")
print("=" * 60 + "\n")

import numpy as np
from sklearn.datasets import load_iris

# Đọc dữ liệu
iris = load_iris()

# Gán X là data và Y là target
X = iris.data
Y = iris.target

# Lấy giá trị lớn nhất theo từng cột (axis=0)
max_cols = X.max(axis=0)

# Chuẩn hóa dữ liệu về đoạn [0, 1]
X_normalized = X / max_cols

print(f"Giá trị lớn nhất của từng cột:\n{max_cols}\n")
print(f"5 dòng dữ liệu X trước khi chuẩn hóa:\n{X[:5]}\n")
print(f"5 dòng dữ liệu X sau khi chuẩn hóa về [0, 1]:\n{X_normalized[:5]}")
```

```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU D: CHIA DỮ LIỆU TẬP TRAIN VÀ TEST (8:2)")
print("=" * 60 + "\n")

from sklearn.model_selection import train_test_split

# Chia tập dữ liệu: test_size=0.2 tương ứng tỷ lệ 8:2
# Sử dụng X_normalized từ câu c và nhãn Y
X_train, X_test, y_train, y_test = train_test_split(
    X_normalized, Y, test_size=0.2, random_state=42, stratify=Y
)

print("Kích thước các tập dữ liệu sau khi chia:")
print(f"- X_train: {X_train.shape} | y_train: {y_train.shape}")
print(f"- X_test : {X_test.shape}  | y_test : {y_test.shape}")
print("\n5 mẫu đầu tiên của X_train:\n", X_train[:5])
print("\n5 nhãn đầu tiên của y_train:\n", y_train[:5])
```

```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU E: MÔ HÌNH CÂY QUYẾT ĐỊNH (DECISION TREE)")
print("=" * 60 + "\n")

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Khởi tạo và huấn luyện mô hình Decision Tree trên tập train
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# 2. Dự đoán nhãn cho tập test
y_pred = dt_model.predict(X_test)
print("Nhãn thực tế (y_test)   :", y_test)
print("Nhãn dự đoán (y_pred)   :", y_pred)
print("-" * 50)

# 3. Tính và in Accuracy trên tập test
acc = accuracy_score(y_test, y_pred)
print(f"\n3. ACCURACY CỦA MÔ HÌNH: {acc:.4f} ({acc * 100:.2f}%)\n")
print("-" * 50)

# 4. Tính và in Precision, Recall, F1-score (classification_report)
print("\n4. BẢNG ĐÁNH GIÁ CHI TIẾT (PRECISION, RECALL, F1-SCORE):")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
print("-" * 50)

# 5. Hiển thị Confusion Matrix bằng Heatmap
print("\n5. MA TRẬN NHẦM LẪN (CONFUSION MATRIX):")
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)
plt.title("Confusion Matrix - Decision Tree\nSV: Vũ Trương Phi - MSSV: 6451071059")
plt.xlabel("Nhãn dự đoán (Predicted Label)")
plt.ylabel("Nhãn thực tế (True Label)")
plt.show()
```

```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU F: MÔ HÌNH NAÏVE BAYES (GAUSSIAN NB)")
print("=" * 60 + "\n")

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Khởi tạo và huấn luyện mô hình GaussianNB
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# 2. Dự đoán nhãn cho tập test
y_pred_nb = nb_model.predict(X_test)
print("Nhãn thực tế (y_test)      :", y_test)
print("Nhãn dự đoán (y_pred_nb)   :", y_pred_nb)
print("-" * 50)

# 3. Tính và in Accuracy trên tập test
acc_nb = accuracy_score(y_test, y_pred_nb)
print(f"\n3. ACCURACY CỦA MÔ HÌNH NAÏVE BAYES: {acc_nb:.4f} ({acc_nb * 100:.2f}%)\n")
print("-" * 50)

# 4. Tính và in Precision, Recall, F1-score (classification_report)
print("\n4. BẢNG ĐÁNH GIÁ CHI TIẾT (PRECISION, RECALL, F1-SCORE):")
print(classification_report(y_test, y_pred_nb, target_names=iris.target_names))
print("-" * 50)

# 5. Hiển thị Confusion Matrix bằng Heatmap
print("\n5. MA TRẬN NHẦM LẪN (CONFUSION MATRIX):")
cm_nb = confusion_matrix(y_test, y_pred_nb)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_nb,
    annot=True,
    fmt='d',
    cmap='Greens',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)
plt.title("Confusion Matrix - Naïve Bayes\nSV: Vũ Trương Phi - MSSV: 6451071059")
plt.xlabel("Nhãn dự đoán (Predicted Label)")
plt.ylabel("Nhãn thực tế (True Label)")
plt.show()
```

```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU G: MÔ HÌNH K-NEAREST NEIGHBORS (KNN)")
print("=" * 60 + "\n")

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Khởi tạo và huấn luyện mô hình KNN (chọn k=5 láng giềng gần nhất mặc định)
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

# 2. Dự đoán nhãn cho tập test
y_pred_knn = knn_model.predict(X_test)
print("Nhãn thực tế (y_test)       :", y_test)
print("Nhãn dự đoán (y_pred_knn)   :", y_pred_knn)
print("-" * 50)

# 3. Tính và in Accuracy trên tập test
acc_knn = accuracy_score(y_test, y_pred_knn)
print(f"\n3. ACCURACY CỦA MÔ HÌNH KNN: {acc_knn:.4f} ({acc_knn * 100:.2f}%)\n")
print("-" * 50)

# 4. Tính và in Precision, Recall, F1-score (classification_report)
print("\n4. BẢNG ĐÁNH GIÁ CHI TIẾT (PRECISION, RECALL, F1-SCORE):")
print(classification_report(y_test, y_pred_knn, target_names=iris.target_names))
print("-" * 50)

# 5. Hiển thị Confusion Matrix bằng Heatmap
print("\n5. MA TRẬN NHẦM LẪN (CONFUSION MATRIX):")
cm_knn = confusion_matrix(y_test, y_pred_knn)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_knn,
    annot=True,
    fmt='d',
    cmap='Oranges',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)
plt.title("Confusion Matrix - KNN\nSV: Vũ Trương Phi - MSSV: 6451071059")
plt.xlabel("Nhãn dự đoán (Predicted Label)")
plt.ylabel("Nhãn thực tế (True Label)")
plt.show()
```

```python
# --- THÔNG TIN SINH VIÊN ---
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : BÀI 1 - CÂU H: PHÂN LỚP BẰNG NEURAL NETWORK")
print("=" * 60 + "\n")

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Thiết lập seed cố định để kết quả tái lập ổn định
tf.random.set_seed(42)
np.random.seed(42)

# 1. Xây dựng Neural Network gồm 02 hidden layers và 01 output layer
model = Sequential([
    # Hidden layer 1: 10 neuron, hàm kích hoạt Relu, input_shape là shape của X_train (4 thuộc tính)
    Dense(10, activation='relu', input_shape=(X_train.shape[1],)),

    # Hidden layer 2: 20 neuron, hàm kích hoạt Relu, fully connected
    Dense(20, activation='relu'),

    # Output layer: số lượng neuron là số class (3 lớp), hàm kích hoạt Softmax
    Dense(len(iris.target_names), activation='softmax')
])

# 2. Biên dịch mô hình với tối ưu hóa adam, loss sparse_categorical_crossentropy, metric accuracy
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("CẤU TRÚC CHI TIẾT MÔ HÌNH NEURAL NETWORK:")
model.summary()
print("-" * 60)

# 3. Huấn luyện mô hình dựa trên tập train: epochs=100, batch_size=1, verbose=1
print("\nTIẾN HÀNH HUẤN LUYỆN (FIT) VỚI 100 EPOCHS:")
model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=1,
    verbose=1
)
print("-" * 60)

# 4. Dự đoán nhãn lớp cho tập test: predict xác suất và dùng np.argmax chọn class có xác suất lớn nhất
y_pred_prob = model.predict(X_test)
y_pred_nn = np.argmax(y_pred_prob, axis=1)

print("\nNhãn thực tế (y_test)         :", y_test)
print("Nhãn dự đoán mô hình (y_pred) :", y_pred_nn)
print("-" * 60)

# 5. Tính và in ra Accuracy của mô hình trên tập test
acc_nn = accuracy_score(y_test, y_pred_nn)
print(f"\n5. ACCURACY TRÊN TẬP TEST: {acc_nn:.4f} ({acc_nn * 100:.2f}%)\n")
print("-" * 60)

# 6. Tính và in ra Precision, Recall, F1-score của từng lớp và trung bình (classification_report)
print("\n6. BẢNG ĐÁNH GIÁ ĐỘ ĐO (PRECISION, RECALL, F1-SCORE):")
print(classification_report(y_test, y_pred_nn, target_names=iris.target_names))
print("-" * 60)

# 7. Hiển thị confusion matrix bằng heat map (bản đồ nhiệt)
print("\n7. BẢN ĐỒ NHIỆT MA TRẬN NHẦM LẪN (CONFUSION MATRIX):")
cm_nn = confusion_matrix(y_test, y_pred_nn)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_nn,
    annot=True,
    fmt='d',
    cmap='Purples',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)
plt.title("Confusion Matrix - Neural Network\nSV: Vũ Trương Phi - MSSV: 6451071059")
plt.xlabel("Nhãn dự đoán (Predicted Label)")
plt.ylabel("Nhãn thực tế (True Label)")
plt.show()
```

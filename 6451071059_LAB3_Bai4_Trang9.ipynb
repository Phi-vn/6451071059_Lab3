```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score
from sklearn.model_selection import KFold
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf
from tensorflow.keras import layers, models

# Thông tin sinh viên
TEN_SINH_VIEN = "Vũ Trương Phi"
MSSV = "6451071059"

print("=" * 60)
print(f"HỌ VÀ TÊN : {TEN_SINH_VIEN}")
print(f"MSSV      : {MSSV}")
print("=" * 60)
```

```python
# a) Đọc bộ dữ liệu Iris từ thư viện sklearn
iris = load_iris()
X = iris.data
y = iris.target

print(f"--- CÂU A: ĐỌC DỮ LIỆU ---")
print(f"Sinh viên thực hiện: {TEN_SINH_VIEN} - {MSSV}")
print(f"Số lượng mẫu (samples): {X.shape[0]}")
print(f"Số lượng đặc trưng (features): {X.shape[1]}")
print(f"Danh sách nhãn: {list(iris.target_names)}")
```

```python
# b) Chuẩn hóa dữ liệu data về đoạn [0,1] sử dụng MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 1))
X_scaled = scaler.fit_transform(X)

print(f"--- CÂU B: CHUẨN HÓA DỮ LIỆU ---")
print(f"Sinh viên thực hiện: {TEN_SINH_VIEN} - {MSSV}")
print(f"Giá trị nhỏ nhất sau chuẩn hóa: {X_scaled.min()}")
print(f"Giá trị lớn nhất sau chuẩn hóa: {X_scaled.max()}")
print(f"5 dòng đầu sau chuẩn hóa:\n{X_scaled[:5]}")
```

```python
# c) Định nghĩa hàm create_model
def create_model():
    model = models.Sequential(
        [
            layers.Input(shape=(4,)),  # Dữ liệu Iris có 4 thuộc tính
            layers.Dense(16, activation="relu"),
            layers.Dense(8, activation="relu"),
            layers.Dense(3, activation="softmax"),  # 3 lớp đầu ra
        ]
    )

    # Biên dịch mô hình theo đúng yêu cầu đề bài
    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model


print(f"--- CÂU C: ĐỊNH NGHĨA HÀM CREATE_MODEL ---")
print(f"Sinh viên thực hiện: {TEN_SINH_VIEN} - {MSSV}")
sample_model = create_model()
sample_model.summary()
```

```python
# d) Huấn luyện k-fold cross validation (k=10), lưu/in metrics và lưu model sau mỗi fold
k = 10
kf = KFold(n_splits=k, shuffle=True, random_state=42)

acc_list, prec_list, rec_list, f1_list = [], [], [], []

print(f"--- CÂU D: HUẤN LUYỆN VÀ ĐÁNH GIÁ K-FOLD (k={k}) ---")
print(f"Sinh viên thực hiện: {TEN_SINH_VIEN} - {MSSV}\n")
print(
    f"{'Fold':<6} | {'Accuracy':<10} | {'Precision':<10} | {'Recall':<10} | {'F1-score':<10}"
)
print("-" * 60)

for fold_idx, (train_idx, val_idx) in enumerate(kf.split(X_scaled, y), start=1):
    X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]

    # Khởi tạo mô hình mới cho mỗi fold
    model = create_model()

    # Huấn luyện mô hình
    model.fit(X_train, y_train, epochs=100, batch_size=8, verbose=0)

    # Dự đoán trên tập validation
    y_pred_probs = model.predict(X_val, verbose=0)
    y_pred = np.argmax(y_pred_probs, axis=1)

    # Tính toán các độ đo đánh giá
    acc = accuracy_score(y_val, y_pred)
    prec = precision_score(y_val, y_pred, average="weighted", zero_division=0)
    rec = recall_score(y_val, y_pred, average="weighted", zero_division=0)
    f1 = f1_score(y_val, y_pred, average="weighted", zero_division=0)

    acc_list.append(acc)
    prec_list.append(prec)
    rec_list.append(rec)
    f1_list.append(f1)

    # Lưu lại mô hình sau mỗi fold
    model.save(f"model_fold_{fold_idx}.keras")

    # In kết quả fold hiện tại
    print(
        f"{fold_idx:<6} | {acc*100:>8.2f}% | {prec*100:>8.2f}% | {rec*100:>8.2f}% | {f1*100:>8.2f}%"
    )

print("-" * 60)
print(f"KẾT QUẢ TRUNG BÌNH {k} FOLDS:")
print(f"Accuracy trung bình  : {np.mean(acc_list)*100:.2f}%")
print(f"Precision trung bình : {np.mean(prec_list)*100:.2f}%")
print(f"Recall trung bình    : {np.mean(rec_list)*100:.2f}%")
print(f"F1-score trung bình  : {np.mean(f1_list)*100:.2f}%")
print("=" * 60)
```

```python
# e) In ra Average Accuracy, Recall, Precision, F1-score toàn cục
avg_acc = np.mean(acc_list)
avg_prec = np.mean(prec_list)
avg_rec = np.mean(rec_list)
avg_f1 = np.mean(f1_list)

print("=" * 65)
print(f"HỌ VÀ TÊN : {TEN_SINH_VIEN} - MSSV: {MSSV}")
print("--- CÂU E: KẾT QUẢ TRUNG BÌNH TOÀN CỤC CỦA MÔ HÌNH (K-FOLD) ---")
print("=" * 65)
print(f"Average Accuracy  : {avg_acc * 100:.2f}%")
print(f"Average Precision : {avg_prec * 100:.2f}%")
print(f"Average Recall    : {avg_rec * 100:.2f}%")
print(f"Average F1-score  : {avg_f1 * 100:.2f}%")
print("=" * 65)
```

```python
# f) Tính và in Average Recall, Precision, F1-score cho từng class
class_names = iris.target_names  # ['setosa', 'versicolor', 'virginica']
n_classes = len(class_names)

prec_per_class = []
rec_per_class = []
f1_per_class = []

# Đánh giá lại trên tập validation của từng fold theo từng class
for fold_idx, (train_idx, val_idx) in enumerate(kf.split(X_scaled, y), start=1):
    X_val = X_scaled[val_idx]
    y_val = y[val_idx]

    # Tải lại mô hình của fold tương ứng
    model = tf.keras.models.load_model(f"model_fold_{fold_idx}.keras")

    y_pred_probs = model.predict(X_val, verbose=0)
    y_pred = np.argmax(y_pred_probs, axis=1)

    # Tính metric riêng cho từng class (average=None trả về mảng cho từng nhãn)
    prec_c = precision_score(
        y_val, y_pred, labels=[0, 1, 2], average=None, zero_division=0
    )
    rec_c = recall_score(
        y_val, y_pred, labels=[0, 1, 2], average=None, zero_division=0
    )
    f1_c = f1_score(
        y_val, y_pred, labels=[0, 1, 2], average=None, zero_division=0
    )

    prec_per_class.append(prec_c)
    rec_per_class.append(rec_c)
    f1_per_class.append(f1_c)

# Tính giá trị trung bình qua các fold cho từng class
avg_prec_c = np.mean(prec_per_class, axis=0)
avg_rec_c = np.mean(rec_per_class, axis=0)
avg_f1_c = np.mean(f1_per_class, axis=0)

print("=" * 65)
print(f"HỌ VÀ TÊN : {TEN_SINH_VIEN} - MSSV: {MSSV}")
print("--- CÂU F: KẾT QUẢ ĐÁNH GIÁ CHI TIẾT THEO TỪNG CLASS ---")
print("=" * 65)
print(
    f"{'Class ID':<10} | {'Class Name':<15} | {'Avg Precision':<15} | {'Avg Recall':<15} | {'Avg F1-score':<15}"
)
print("-" * 75)
for i in range(n_classes):
    print(
        f"{i:<10} | {class_names[i]:<15} | {avg_prec_c[i]*100:>12.2f}% | {avg_rec_c[i]*100:>12.2f}% | {avg_f1_c[i]*100:>12.2f}%"
    )
print("=" * 75)
```

```python
# g) Dự đoán 3 mẫu dữ liệu mới bằng (các) mô hình có F1-score tốt nhất
new_samples = np.array(
    [[6.2, 2.9, 4.3, 1.3], [5.1, 3.5, 1.4, 0.2], [7.3, 2.8, 6.4, 2.1]]
)

# Chuẩn hóa dữ liệu mới theo scaler đã fit từ trước
new_samples_scaled = scaler.transform(new_samples)

# Tìm F1-score cao nhất và lấy danh sách các fold đạt F1-score tốt nhất đó
max_f1 = max(f1_list)
best_fold_indices = [
    i + 1 for i, score in enumerate(f1_list) if np.isclose(score, max_f1)
]

print("=" * 70)
print(f"HỌ VÀ TÊN : {TEN_SINH_VIEN} - MSSV: {MSSV}")
print("--- CÂU G: DỰ ĐOÁN 03 MẪU DỮ LIỆU MỚI ---")
print(f"[+] F1-score cao nhất đạt được: {max_f1 * 100:.2f}%")
print(f"[+] Các fold có mô hình tốt nhất: Fold {best_fold_indices}")
print("=" * 70)

# Dự đoán bằng từng mô hình tốt nhất
for fold_num in best_fold_indices:
    best_model = tf.keras.models.load_model(f"model_fold_{fold_num}.keras")
    preds = best_model.predict(new_samples_scaled, verbose=0)
    pred_classes = np.argmax(preds, axis=1)

    print(
        f"\n>>> Kết quả dự đoán từ mô hình Fold {fold_num} (F1-score: {max_f1*100:.2f}%):"
    )
    for idx, (sample, class_id) in enumerate(zip(new_samples, pred_classes), 1):
        label_name = class_names[class_id]
        prob = preds[idx - 1][class_id] * 100
        print(
            f"  - Mẫu {idx} {sample} => Lớp {class_id} ({label_name}) với độ tin cậy {prob:.2f}%"
        )
print("=" * 70)
```

```python
# ==============================================================================
# CÂU a: ĐỌC BỘ DỮ LIỆU TRAIN, TEST CỦA MNIST TỪ KERAS
# ==============================================================================
import tensorflow as tf
from tensorflow.keras.datasets import mnist

# In thông tin sinh viên
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3a: Tải bộ dữ liệu MNIST")
print("=" * 60)

# Tải bộ dữ liệu MNIST từ thư viện Keras
(X_train, y_train), (X_test, y_test) = mnist.load_data()

print(">> Đã tải thành công bộ dữ liệu MNIST từ Keras!")
```

```python
# ==============================================================================
# CÂU b: IN RA KÍCH THƯỚC CỦA TẬP TRAIN, TEST
# ==============================================================================

# In thông tin sinh viên
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3b: Kích thước tập Train và Test")
print("=" * 60)

# In kích thước các tập
print(f"Kích thước tập ảnh Train (X_train) : {X_train.shape}")
print(f"Kích thước tập nhãn Train (y_train): {y_train.shape}")
print(f"Kích thước tập ảnh Test  (X_test)  : {X_test.shape}")
print(f"Kích thước tập nhãn Test (y_test) : {y_test.shape}")
```

```python
# ==============================================================================
# CÂU c: CHUẨN HÓA CÁC GIÁ TRỊ PIXEL CỦA ẢNH VỀ ĐOẠN [0, 1]
# ==============================================================================

# In thông tin sinh viên
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3c: Chuẩn hóa dữ liệu pixel ảnh")
print("=" * 60)

# Chuyển kiểu dữ liệu sang float32 và chia cho 255.0 để đưa về [0, 1]
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Kiểm tra lại giá trị Min và Max sau khi chuẩn hóa
print(f">> Giá trị pixel nhỏ nhất (Min): {X_train.min()}")
print(f">> Giá trị pixel lớn nhất (Max): {X_train.max()}")
print(">> Chuẩn hóa dữ liệu về đoạn [0, 1] thành công!")
```

```python
# ==============================================================================
# CÂU d: ĐỊNH NGHĨA HÀM create_model
# ==============================================================================
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

# In thông tin sinh viên
print("=" * 60)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3d: Định nghĩa hàm create_model")
print("=" * 60)

def create_model():
    # Xây dựng kiến trúc Neural Network tương tự Câu 2
    model = Sequential([
        Flatten(input_shape=(28, 28)),          # Duỗi ảnh 28x28 thành vector 784
        Dense(128, activation='relu'),           # Hidden layer với 128 nơ-ron
        Dense(10, activation='softmax')          # Output layer với 10 nơ-ron (cho các số 0-9)
    ])

    # Biên dịch mô hình theo đúng yêu cầu đề bài
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# Khởi tạo thử mô hình và in bản tóm tắt kiến trúc mạng
sample_model = create_model()
sample_model.summary()
```

```python
# ==============================================================================
# CÂU e: HUẤN LUYỆN MÔ HÌNH BẰNG K-FOLD CROSS VALIDATION (K = 10)
# ==============================================================================
import numpy as np
from sklearn.model_selection import KFold

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3e: K-Fold Cross Validation (k=10)")
print("=" * 65, "\n")

# Định nghĩa kfold với n_splits = 10, shuffle = True, random_state = 42
kfold = KFold(n_splits=10, shuffle=True, random_state=42)

# Các danh sách để lưu trữ mô hình và kết quả sau mỗi fold
models = []
loss_per_fold = []
accuracy_per_fold = []

fold_num = 1

# Vòng lặp duyệt qua từng fold
for train_index, val_index in kfold.split(X_train):
    print(f"--- Đang huấn luyện Fold {fold_num}/10 ---")

    # 1. Gán model là kết quả của hàm create_model()
    model = create_model()

    # 2. Gán X_train_fold, X_val_fold, y_train_fold, y_val_fold
    X_train_fold, X_val_fold = X_train[train_index], X_train[val_index]
    y_train_fold, y_val_fold = y_train[train_index], y_train[val_index]

    # 3. Huấn luyện mô hình: epoch: 5; batch_size: 32
    history = model.fit(
        X_train_fold, y_train_fold,
        epochs=5,
        batch_size=32,
        validation_data=(X_val_fold, y_val_fold),
        verbose=1
    )

    # 4. Lưu model huấn luyện được vào mảng models
    models.append(model)

    # 5. Tính và in accuracy, loss của fold
    scores = model.evaluate(X_val_fold, y_val_fold, verbose=0)
    loss = scores[0]
    acc = scores[1]

    # Lưu vào mảng
    loss_per_fold.append(loss)
    accuracy_per_fold.append(acc)

    print(f">> Kết quả Fold {fold_num}: Loss = {loss:.4f} - Accuracy = {acc * 100:.2f}%\n")
    fold_num += 1

# Hiển thị tổng kết 10 folds
print("=" * 65)
print("              BẢNG TỔNG HỢP KẾT QUẢ 10 FOLDS")
print("=" * 65)
for i in range(len(accuracy_per_fold)):
    print(f"Fold {i+1:2d}: Loss = {loss_per_fold[i]:.4f} | Accuracy = {accuracy_per_fold[i] * 100:.2f}%")

print("-" * 65)
print(f"Độ chính xác trung bình (Mean Accuracy): {np.mean(accuracy_per_fold) * 100:.2f}%")
print(f"Độ mất mát trung bình (Mean Loss)      : {np.mean(loss_per_fold):.4f}")
print("=" * 65)
```

```python
# ==============================================================================
# CÂU f: TÍNH VÀ IN MEAN & ĐỘ LỆCH CHUẨN CỦA ACCURACY VÀ LOSS
# ==============================================================================
import numpy as np

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3f: Thống kê Accuracy và Loss qua các Folds")
print("=" * 65)

# Sử dụng mean và std trong numpy theo hướng dẫn
mean_acc = np.mean(accuracy_per_fold)
std_acc = np.std(accuracy_per_fold)
mean_loss = np.mean(loss_per_fold)
std_loss = np.std(loss_per_fold)

print(f">> Mean Accuracy (Độ chính xác trung bình) : {mean_acc * 100:.2f}%")
print(f">> Standard Deviation Accuracy (Độ lệch chuẩn Acc): {std_acc * 100:.2f}%")
print(f">> Mean Loss (Độ mất mát trung bình)      : {mean_loss:.4f}")
print(f">> Standard Deviation Loss (Độ lệch chuẩn Loss)  : {std_loss:.4f}")
print("=" * 65)
```

```python
# ==============================================================================
# CÂU g: VẼ BIỂU ĐỒ THỂ HIỆN ACCURACY VÀ LOSS SAU MỖI FOLD
# ==============================================================================
import matplotlib.pyplot as plt

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3g: Biểu đồ Accuracy và Loss")
print("=" * 65)

folds = list(range(1, len(accuracy_per_fold) + 1))

plt.figure(figsize=(14, 5))

# 1. Biểu đồ Accuracy
plt.subplot(1, 2, 1)
plt.plot(folds, accuracy_per_fold, marker='o', color='b', linewidth=2, label='Accuracy')
plt.title('Accuracy qua từng Fold (Vũ Trương Phi - 6451071059)', fontsize=12)
plt.xlabel('Fold', fontsize=11)
plt.ylabel('Accuracy', fontsize=11)
plt.xticks(folds)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# 2. Biểu đồ Loss
plt.subplot(1, 2, 2)
plt.plot(folds, loss_per_fold, marker='s', color='r', linewidth=2, label='Loss')
plt.title('Loss qua từng Fold (Vũ Trương Phi - 6451071059)', fontsize=12)
plt.xlabel('Fold', fontsize=11)
plt.ylabel('Loss', fontsize=11)
plt.xticks(folds)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# CÂU h: CHỌN MÔ HÌNH TỐT NHẤT VÀ ĐÁNH GIÁ TRÊN TẬP TEST
# ==============================================================================
import numpy as np

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3h: Đánh giá mô hình tốt nhất trên tập Test")
print("=" * 65)

# Tìm index có accuracy lớn nhất trong accuracy_per_fold bằng argmax
best_fold_idx = np.argmax(accuracy_per_fold)
best_model = models[best_fold_idx]

print(f">> Fold có kết quả tốt nhất: Fold {best_fold_idx + 1}")
print(f">> Accuracy của fold này lúc huấn luyện: {accuracy_per_fold[best_fold_idx] * 100:.2f}%\n")

# Đánh giá mô hình tốt nhất trên tập dữ liệu Test độc lập (X_test, y_test)
test_loss, test_acc = best_model.evaluate(X_test, y_test, verbose=1)

print("\n" + "=" * 65)
print("              KẾT QUẢ PHÂN LỚP TRÊN TẬP TEST (MNIST)")
print("=" * 65)
print(f">> Test Loss     : {test_loss:.4f}")
print(f">> Test Accuracy : {test_acc * 100:.2f}%")
print("=" * 65)
print("Hoàn thành toàn bộ Bài 3 Lab 3!")
print("=" * 65)
```

```python
# ==============================================================================
# CÂU i: TÍNH VÀ IN RA ACCURACY CỦA MÔ HÌNH TRÊN TẬP TEST
# ==============================================================================
import numpy as np
from sklearn.metrics import accuracy_score

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3i: Accuracy trên tập Test")
print("=" * 65)

# Dự đoán xác suất cho toàn bộ tập test bằng mô hình tốt nhất (best_model)
y_pred_probs = best_model.predict(X_test)

# Chuyển xác suất sang nhãn dự đoán (lớp có xác suất cao nhất)
y_pred = np.argmax(y_pred_probs, axis=1)

# Tính độ chính xác Accuracy bằng accuracy_score
test_accuracy = accuracy_score(y_test, y_pred)

print(f"\n>> ĐỘ CHÍNH XÁC (ACCURACY) TRÊN TẬP TEST: {test_accuracy * 100:.2f}%")
print("=" * 65)
```

```python
# ==============================================================================
# CÂU j: TÍNH PRECISION, RECALL, F1-SCORE CỦA TỪNG LỚP VÀ TRUNG BÌNH
# ==============================================================================
from sklearn.metrics import classification_report

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3j: Classification Report trên tập Test")
print("=" * 65)

# In báo cáo chi tiết gồm Precision, Recall, F1-score cho từng lớp (0 -> 9) và trung bình
report = classification_report(
    y_test,
    y_pred,
    target_names=[f"Chữ số {i}" for i in range(10)],
    digits=4
)

print("\nBÁO CÁO ĐÁNH GIÁ CHI TIẾT HIỆU NĂNG MÔ HÌNH:")
print(report)
print("=" * 65)
```

```python
# ==============================================================================
# CÂU k: HIỂN THỊ CONFUSION MATRIX BẰNG HEAT MAP
# ==============================================================================
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# In thông tin sinh viên
print("=" * 65)
print("HỌ VÀ TÊN : VŨ TRƯƠNG PHI")
print("MSSV      : 6451071059")
print("BÀI TẬP   : LAB 3 - CÂU 3k: Confusion Matrix Heatmap")
print("=" * 65)

# Tính ma trận nhầm lẫn
cm = confusion_matrix(y_test, y_pred)

# Vẽ biểu đồ nhiệt (Heatmap)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=True,
            xticklabels=[str(i) for i in range(10)],
            yticklabels=[str(i) for i in range(10)])

plt.title('Confusion Matrix Heatmap (Vũ Trương Phi - 6451071059)', fontsize=13, pad=15)
plt.xlabel('Nhãn Dự Đoán (Predicted Label)', fontsize=11)
plt.ylabel('Nhãn Thực Tế (True Label)', fontsize=11)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CÂU A: ĐỌC DỮ LIỆU TỪ KERAS
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
from tensorflow.keras.datasets import mnist

# Đọc bộ dữ liệu và lưu vào các biến theo đúng hướng dẫn
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-> Đã đọc xong bộ dữ liệu train, test của MNIST từ Keras thành công!")
```

```python
# ==========================================
# CÂU B: IN RA KÍCH THƯỚC (SHAPE)
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 50)
print(f"Kích thước tập train_images : {train_images.shape}")
print(f"Kích thước tập train_labels : {train_labels.shape}")
print(f"Kích thước tập test_images  : {test_images.shape}")
print(f"Kích thước tập test_labels  : {test_labels.shape}")
```

```python
# ==========================================
# CÂU C: CHUẨN HÓA DỮ LIỆU VỀ ĐOẠN [0, 1]
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
# Ép kiểu sang float32 và chia cho 255.0 theo hướng dẫn
train_images = train_images.astype("float32") / 255.0
test_images = test_images.astype("float32") / 255.0

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 50)
print(f"Giá trị pixel nhỏ nhất sau khi chuẩn hóa : {train_images.min()}")
print(f"Giá trị pixel lớn nhất sau khi chuẩn hóa : {train_images.max()}")
print("-> Chuẩn hóa pixel ảnh về đoạn [0, 1] hoàn tất!")
```

```python
# ==========================================
# CÂU D: CHIA TẬP TRAIN VÀ VALIDATION (8:2)
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
from sklearn.model_selection import train_test_split

# Chia tập ban đầu với test_size=0.2 (tương ứng tỷ lệ 8:2)
(
    train_images,
    validation_images,
    train_labels,
    validation_labels,
) = train_test_split(
    train_images, train_labels, test_size=0.2, random_state=42
)

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 50)
print("KÍCH THƯỚC CÁC TẬP DỮ LIỆU:")
print(f"- Kích thước tập train_images      : {train_images.shape}")
print(f"- Kích thước tập train_labels      : {train_labels.shape}")
print(f"- Kích thước tập validation_images : {validation_images.shape}")
print(f"- Kích thước tập validation_labels : {validation_labels.shape}")
```

```python
# =========================================================
# CÂU E: XÂY DỰNG MÔ HÌNH NEURAL NETWORK
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# =========================================================
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense, Dropout

# Khởi tạo mô hình Sequential
model = Sequential([
    Flatten(input_shape=(28, 28)),          # Chuyển ảnh 28x28 thành vector 784
    Dense(128, activation='relu'),          # Hidden layer gồm 128 neural, kích hoạt Relu
    Dropout(0.2),                           # Dropout layer với xác suất 20%
    Dense(10, activation='softmax')         # Output layer gồm 10 neural (10 class từ 0-9), Softmax
])

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 55)
print("KIẾN TRÚC MÔ HÌNH (SUMMARY):")
model.summary()
```

```python
# =========================================================
# CÂU F: BIÊN DỊCH MÔ HÌNH
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# =========================================================
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 55)
print("-> Đã biên dịch mô hình thành công với:")
print("   + Optimizer: adam")
print("   + Loss: sparse_categorical_crossentropy")
print("   + Metric: accuracy")
```

```python
# =========================================================
# CÂU G: HUẤN LUYỆN MÔ HÌNH (TRAIN & VALIDATION)
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# =========================================================
print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 55)
print("BẮT ĐẦU HUẤN LUYỆN MÔ HÌNH (EPOCHS=5, BATCH_SIZE=32):")

history = model.fit(
    train_images,
    train_labels,
    epochs=5,
    batch_size=32,
    validation_data=(validation_images, validation_labels),
    verbose=1
)

print("-" * 55)
print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-> Hoàn thành quá trình huấn luyện mô hình!")
```

```python
# ==========================================
# CÂU H: VẼ BIỂU ĐỒ ACCURACY VÀ LOSS
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
import matplotlib.pyplot as plt

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 55)

epochs_range = range(1, len(history.history['accuracy']) + 1)

plt.figure(figsize=(14, 5))

# 1. Biểu đồ Accuracy
plt.subplot(1, 2, 1)
plt.plot(epochs_range, history.history['accuracy'], 'b-o', label='Train Accuracy')
plt.plot(epochs_range, history.history['val_accuracy'], 'r--s', label='Validation Accuracy')
plt.title('Độ chính xác qua từng Epoch (Vũ Trương Phi - 6451071059)')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# 2. Biểu đồ Loss
plt.subplot(1, 2, 2)
plt.plot(epochs_range, history.history['loss'], 'b-o', label='Train Loss')
plt.plot(epochs_range, history.history['val_loss'], 'r--s', label='Validation Loss')
plt.title('Độ mất mát qua từng Epoch (Vũ Trương Phi - 6451071059)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CÂU I: DỰ ĐOÁN PHÂN LỚP CHO TẬP TEST
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
import numpy as np

# Dự đoán xác suất cho mỗi mẫu và mỗi class bằng hàm predict
predictions_prob = model.predict(test_images)

# Dùng np.argmax để chọn ra class có xác suất lớn nhất
y_pred = np.argmax(predictions_prob, axis=1)

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 55)
print(f"-> Kích thước mảng dự đoán xác suất : {predictions_prob.shape}")
print(f"-> Số lượng nhãn dự đoán cho tập test: {len(y_pred)}")
print(f"-> Dự đoán cho 10 mẫu đầu tiên       : {y_pred[:10]}")
print(f"-> Nhãn thực tế của 10 mẫu đầu tiên  : {test_labels[:10]}")
```

```python
# ==========================================
# CÂU J: TÍNH VÀ IN RA ACCURACY TRÊN TẬP TEST
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==========================================
import numpy as np
from sklearn.metrics import accuracy_score

# Cách 1: Tính accuracy trực tiếp từ kết quả argmax ở câu i
test_acc_from_pred = accuracy_score(test_labels, y_pred)

# Cách 2: Đánh giá trực tiếp qua hàm evaluate của model
test_loss, test_acc_eval = model.evaluate(test_images, test_labels, verbose=0)

print("=" * 60)
print(f"KẾT QUẢ ĐÁNH GIÁ MÔ HÌNH TRÊN TẬP TEST")
print(f"Sinh viên: Vũ Trương Phi - MSSV: 6451071059")
print("=" * 60)
print(f"-> Test Loss                            : {test_loss:.4f}")
print(f"-> Test Accuracy (tính bằng accuracy_score): {test_acc_from_pred * 100:.2f}%")
print(f"-> Test Accuracy (tính bằng model.evaluate): {test_acc_eval * 100:.2f}%")
print("=" * 60)
```

```python
# ==============================================================================
# CÂU K: TÍNH PRECISION, RECALL, F1-SCORE TRÊN TẬP TEST
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==============================================================================
from sklearn.metrics import classification_report

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 65)
print("BÁO CÁO PHÂN LỚP CHI TIẾT (CLASSIFICATION REPORT):")

# In báo cáo chi tiết từng lớp từ 0 đến 9 và các giá trị trung bình (macro avg, weighted avg)
target_names = [f"Chữ số {i}" for i in range(10)]
report = classification_report(test_labels, y_pred, target_names=target_names, digits=4)
print(report)
```

```python
# ==============================================================================
# CÂU L: HIỂN THỊ CONFUSION MATRIX BẰNG HEATMAP
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==============================================================================
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 65)

# Tính ma trận nhầm lẫn
cm = confusion_matrix(test_labels, y_pred)

# Vẽ heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=True,
            xticklabels=range(10), yticklabels=range(10))
plt.title('Ma trận nhầm lẫn (Confusion Matrix) - Vũ Trương Phi (6451071059)', fontsize=12)
plt.xlabel('Nhãn dự đoán (Predicted Label)', fontsize=11)
plt.ylabel('Nhãn thực tế (True Label)', fontsize=11)
plt.show()
```

```python
# ==============================================================================
# CÂU M: DỰ ĐOÁN VÀ HIỂN THỊ 05 ẢNH ĐẦU TIÊN TẬP TEST
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 65)

plt.figure(figsize=(15, 3))

for idx in range(5):
    # Lấy ảnh thứ idx và dùng expand_dims để thêm chiều batch (1, 28, 28)
    img = test_images[idx]
    img_expanded = np.expand_dims(img, axis=0)

    # Dự đoán xác suất và chọn class có xác suất cao nhất bằng argmax
    prob = model.predict(img_expanded, verbose=0)
    pred_label = np.argmax(prob)
    true_label = test_labels[idx]

    # In kết quả dạng text
    print(f"Ảnh index {idx}: Dự đoán = {pred_label} | Thực tế = {true_label}")

    # Hiển thị ảnh kèm nhãn
    plt.subplot(1, 5, idx + 1)
    plt.imshow(img, cmap='gray')
    plt.title(f"Pred: {pred_label}\nTrue: {true_label}", color="green" if pred_label == true_label else "red")
    plt.axis('off')

plt.suptitle("Phân lớp 05 ảnh đầu tiên (Vũ Trương Phi - 6451071059)", y=1.05, fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# CÂU N: IN RA 05 ẢNH ĐẦU TIÊN BỊ PHÂN LỚP SAI
# Sinh viên: Vũ Trương Phi - MSSV: 6451071059
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt

print(f"Họ và tên: Vũ Trương Phi | MSSV: 6451071059")
print("-" * 65)

# Tìm các chỉ số index mà nhãn dự đoán khác nhãn thực tế
wrong_indices = np.where(y_pred != test_labels)[0]

print(f"-> Tổng số ảnh bị phân lớp sai trên tập test: {len(wrong_indices)} / {len(test_labels)}")
print("-> Hiển thị 05 trường hợp phân lớp sai đầu tiên:")

# Lấy 5 trường hợp sai đầu tiên
first_5_wrong = wrong_indices[:5]

plt.figure(figsize=(15, 3))

for i, idx in enumerate(first_5_wrong):
    img = test_images[idx]
    pred_label = y_pred[idx]
    true_label = test_labels[idx]

    print(f"  + Mẫu test index {idx}: Thực tế là '{true_label}' nhưng mô hình đoán nhầm thành '{pred_label}'")

    # Hiển thị hình ảnh minh họa mẫu đoán sai
    plt.subplot(1, 5, i + 1)
    plt.imshow(img, cmap='gray')
    plt.title(f"Index: {idx}\nPred: {pred_label} (Sai)\nTrue: {true_label}", color="red")
    plt.axis('off')

plt.suptitle("05 ảnh đầu tiên bị phân lớp sai (Vũ Trương Phi - 6451071059)", y=1.08, fontsize=13)
plt.tight_layout()
plt.show()
```
